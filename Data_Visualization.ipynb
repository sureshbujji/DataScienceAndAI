```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew

# Generate data
np.random.seed(42)
normal_data = np.random.normal(loc=0, scale=1, size=1000)          # Normal
right_skewed = np.random.exponential(scale=1, size=1000)           # Right-skewed
left_skewed = -np.random.exponential(scale=1, size=1000)           # Left-skewed (invert sign)

# Calculate skewness
print("Normal skewness:", skew(normal_data))
print("Right-skewed skewness:", skew(right_skewed))
print("Left-skewed skewness:", skew(left_skewed))

# Plot
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.hist(left_skewed, bins=30, color='skyblue', edgecolor='black')
plt.title("Left-Skewed (Negative)\nMean < Median")

plt.subplot(1, 3, 2)
plt.hist(normal_data, bins=30, color='lightgreen', edgecolor='black')
plt.title("Normal Distribution\nMean = Median")

plt.subplot(1, 3, 3)
plt.hist(right_skewed, bins=30, color='salmon', edgecolor='black')
plt.title("Right-Skewed (Positive)\nMean > Median")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew

# Generate data
np.random.seed(42)
normal_data = np.random.normal(0, 1, 1000)
right_skewed = np.random.exponential(1, 1000)
left_skewed = -np.random.exponential(1, 1000)

datasets = {
    "Left-Skewed": left_skewed,
    "Normal": normal_data,
    "Right-Skewed": right_skewed
}

# Plot setup
plt.figure(figsize=(15, 10))

# 1️⃣ Histogram + KDE
for i, (title, data) in enumerate(datasets.items(), 1):
    plt.subplot(3, 4, i)
    sns.histplot(data, kde=True, color='skyblue')
    plt.title(f"{title}\nHistogram + KDE")

# 2️⃣ Boxplot
for i, (title, data) in enumerate(datasets.items(), 1):
    plt.subplot(3, 4, i + 3)
    sns.boxplot(x=data, color='lightgreen')
    plt.title(f"{title}\nBoxplot")

# 3️⃣ Violin Plot
for i, (title, data) in enumerate(datasets.items(), 1):
    plt.subplot(3, 4, i + 6)
    sns.violinplot(x=data, color='salmon')
    plt.title(f"{title}\nViolin Plot")

# 4️⃣ Density Curve Overlay (KDE only)
for i, (title, data) in enumerate(datasets.items(), 1):
    plt.subplot(3, 4, i + 9)
    sns.kdeplot(data, fill=True)
    plt.title(f"{title}\nKDE (Density Curve)")

plt.tight_layout()
plt.show()

# Print skewness values
for name, data in datasets.items():
    print(f"{name} skewness: {skew(data):.2f}")
```

```python
import numpy as np
from scipy.stats import kurtosis

# Example data
normal = np.random.normal(0, 1, 1000)
leptokurtic = np.random.laplace(0, 1, 1000)  # heavy tails
platykurtic = np.random.uniform(-3, 3, 1000) # light tails

print("Normal kurtosis:", kurtosis(normal))
print("Leptokurtic kurtosis:", kurtosis(leptokurtic))
print("Platykurtic kurtosis:", kurtosis(platykurtic))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Sample data
data = pd.DataFrame({
    'Year': [2018, 2019, 2020, 2021, 2022],
    'Sales': [100, 150, 130, 180, 200],
    'Profit': [20, 25, 18, 30, 35]
})

plt.figure(figsize=(10,5))
sns.lineplot(x='Year', y='Sales', data=data, marker='o', label='Sales')
sns.lineplot(x='Year', y='Profit', data=data, marker='o', label='Profit')
plt.title("Sales & Profit Trend Over Years")
plt.legend()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Generate example data
np.random.seed(42)
data = pd.DataFrame({
    "Group A": np.random.normal(60, 10, 100),
    "Group B": np.random.normal(50, 5, 100),
    "Group C": np.random.normal(70, 15, 100)
})

# Plot box plot
plt.figure(figsize=(8,5))
sns.boxplot(data=data, palette="Set2")
plt.title("Box Plot Comparison of 3 Groups", fontsize=14)
plt.ylabel("Values")
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Generate right-skewed data (Exponential distribution)
np.random.seed(42)
right_skewed = np.random.exponential(scale=2, size=1000)

# Create box plot
plt.figure(figsize=(7,4))
sns.boxplot(x=right_skewed, color='skyblue')
plt.title("Box Plot of Right-Skewed Distribution", fontsize=14)
plt.xlabel("Value")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
left_skewed = -np.random.exponential(scale=2, size=1000)
normal_data = np.random.normal(0, 1, 1000)

sns.boxplot(data=[left_skewed, normal_data, right_skewed],
            orient="h", palette="Set2")
plt.yticks([0,1,2], ["Left-Skewed", "Normal", "Right-Skewed"])
plt.title("Comparison of Skew Types Using Box Plots", fontsize=14)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams['figure.figsize'] = 6,2
import warnings
warnings.filterwarnings('ignore')
```

```python
stats['InternetUsers']
```

```python
vis1=sns.distplot(stats['InternetUsers'])
plt.show()
```

BOX PLOT

```python
vis4=sns.boxplot(data=stats,x='IncomeGroup',y='BirthRate')
plt.show()
```

# Linear model plot

```python
vis5=sns.lmplot(data=stats,x='InternetUsers',y='BirthRate',fit_reg=False)
plt.show()
```

```python
vis5=sns.lmplot(data=stats,x='InternetUsers',y='BirthRate',fit_reg=True)
plt.show()
```

Hue for color seperation

```python
vis5=sns.lmplot(data=stats,x='InternetUsers',y='BirthRate',fit_reg=True,hue='IncomeGroup')
plt.show()
```
